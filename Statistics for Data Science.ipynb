# Hypothesis Testing
## Z-Test

Example 1:-
A teacher claims that the mean score of students in his class is greater than 82 with a standard deviation of 20. If a simple of 81 students was 
selected with a mean score of 90.

```python
s_mean = 90
p_mean = 82
p_std = 20
n = 81
alpha = 0.05
```

```python
z_cal = (s_mean - p_mean)/ (p_std/np.sqrt(n))
z_cal
```

```python
z_table = st.norm.ppf(1-alpha)
z_table
```

```python
if z_table<z_cal:
    print("Ha is right")
else:
    print("Ho is right")
```

```python
old_design_data = np.array([45.2,42.8,38.9, 43.3,41.3,42.6,40.8,43.2,45.2,46.7,42.4,41.2,40.3,40.5,43.2,45.1,44.2,43.2,39.4,38.7,41.4,46.3,42.5,41.8,45.3,43.9,45.2,42.7,44.1,44.7])
new_design_data = np.array([48.5,46.3,47.5,49.2,48.9,50.3,49.2,48.7,50.4,49.2,50.6,46.3,47.4,48.3,49.9,48.2,46.3,47.7,48.4,50.4,49.2,48.1,49.3,46.2,47.3,48.4,50.4,49.2,47.6,46.8])
```

```python
mean_new = np.mean(new_design_data)
mean_old = np.mean(old_design_data)
pop_std = 2.5
n_sm = len(new_design_data)
apl = 0.05
```

```python
z_cal = (mean_new - mean_old) / (pop_std/np.sqrt(n_sm))
z_cal
```

```python
z_table = st.norm.ppf(1-apl)
z_table
```

```python
if z_table < z_cal:
    print("Ha is right")
else:
    print("Ho is right")
```

A manufacturer claims that the average weight if a bag of potato chips is 150 grams. A sample of 25 bags is taken, and A grams with a standard 
deviation of 5 grams. Test the manufacturers clains using a one-triled with a significance level of 0.05.

```python
t_table = st.t.ppf(0.05,24)
t_table
```

```python
p_mean = 150
s_mean = 148
s_std = 5
s_no =25
```

A company wants to test whether there is a difference in productivity between two teams. They randomly select 20 employees from each team and record 
their prductivity scores. The mean productivity score for Team A is 80 with a standard deviation of 5, while the mran productibity score for team B is 75 with a stanard deviation of 6. Test at a 5% level if significance there id a fifference productivity between the two teams.

Example 3:-
A company wants to test whether a new training program improves the typing speed of its employees. The typing speed of 20 employees was recorded before and after the training program. The data is given below. Test at a 5% level of significance whether the training progtam has an effect on the typing speed of the employees.

```python
before_data = np.array([50,60,45,65,55,70,40,75,80,65,70,60,50,55,45,75,60,50,65,70])
after_data = np.array([60,70,55,75,65,80,50,85,90,70,75,65,55,60,50,80,65,55,70,75])
```

```python
after_mean = np.mean(after_data)
before_mean = np.mean(before_data)
after_std = np.std(after_data)
before_std = np.std(before_data)
after_n = len(after_data)
before_n = len(before_data)
```

```python
t_table = st.t.ppf(1-0.025, (after_n-1))
t_table
```

```python
t_cal = (after_mean - before_mean)/np.sqrt(((np.square(after_std) / after_n))+((np.square(before_std) / before_n)))
t_cal
```

```python
if t_table < t_cal:
    print("Ha is right")
else:
    print("Ho is right")
```

Example 2:-
A study was conducted to investigate whether there is a relationship between gender and preferred genre of music. A sample of 235 people was selected, and the data collected is shown below. Test at a 5% level of significance whether there is a significant association between gender and music preference.

```python
row1 = np.array([40,45,25,10])
row2 = np.array([35,30,20,30])
```

```python
sum_r1 = np.sum(row1)
sum_r2 = np.sum(row2)
sum_row = np.array([sum_r1,sum_r2])
sum_row
```

```python
sum_cal = row1+row2
sum_cal
```

```python
exp = []
for i in sum_row:
    for j in sum_cal:
        value= (i*j)/235
        exp.append(value)
```

```python
obj = np.concatenate([row1,row2])
obj
```

```python
chi_cal = np.sum(np.square(obj-exp)/ exp)
chi_cal
```

```python
chi_table = 7.815
```

```python
if chi_cal > chi_table:
    print("Ha Association")
else:
    print("Ho No Association")
```
