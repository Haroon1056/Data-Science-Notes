```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("personal_information_data.csv")
```

```python
df
```

# Measures of Central Tendency

```python
df["Age"].sum()/len(df["Age"])
```

```python
df["Age"].mean()
```

```python
mn =np.mean(df["Age"])
md = np.median(df["Age"])
mo = df["Age"].mode()[0]
```

```python
sns.histplot(x="Age", data=df)
plt.plot([mn for i in range(0,25)], [i for i in range(0,25)], color='r', label="mean")
plt.plot([md for i in range(0,25)], [i for i in range(0,25)], color='g', label="median")
plt.plot([mo for i in range(0,25)], [i for i in range(0,25)], color='b', label="mode")
plt.legend()
plt.show()
```

# Mean Absolute Deviation

```python
sec_a = np.array([75, 65, 73, 68, 72, 67])
sec_b = np.array([90, 47, 43, 96, 93, 51])
no = np.array([1,2,3,4,5,6])
```

```python
mad = np.sum(np.abs(sec_a-np.mean(sec_a))/len(sec_a))
```

```python
mad
```

```python
mean = np.mean(sec_b)
```

```python
plt.figure(figsize=(10,3))
plt.scatter(sec_a, no, label="sec_a")
plt.scatter(sec_b, no, c="red", label="sec_b")
plt.plot([mean for i in range(1,7)], no, color="green", label="mean")
plt.legend()
plt.show()
```
